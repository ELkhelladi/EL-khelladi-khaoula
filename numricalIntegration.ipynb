```python

import math


def f(x):
    return math.exp(-x**2)



def trapezoidal(a, b, n):
    h = (b - a) / n
    S = (f(a) + f(b)) / 2

    for i in range(1, n):
        x = a + i * h
        S += f(x)

    return h * S



def simpson(a, b, n):
    if n % 2 != 0:
        raise ValueError("n doit être pair")

    h = (b - a) / n
    S = f(a) + f(b)

    for i in range(1, n):
        x = a + i * h

        if i % 2 == 0:
            S += 2 * f(x)
        else:
            S += 4 * f(x)

    return (h / 3) * S



a = 0
b = 1



for n in [4, 6]:
    print("n =", n)
    print("Trapèzes :", trapezoidal(a, b, n))
    print("Simpson  :", simpson(a, b, n))
    print()
```
